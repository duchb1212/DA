Lương trong ngành IT luôn là chủ đề được quan tâm nhưng thiếu phân tích dữ liệu cụ thể. Đề tài này nhằm trả lời **4 câu hỏi nghiên cứu** sau:

| # | Câu hỏi | Tầng phân tích |

## 2. 📂 DATASET

### Các cột hiện có trong dataset

| Cột | Kiểu dữ liệu | Mô tả |
|-----|-------------|-------|
| `rating` | float | Điểm đánh giá công ty |
| `company_name` | object | Tên công ty |
| `job_title` | object | Chức danh công việc |
| `salary` | object → float | Mức lương (USD) — cần làm sạch |
| `salaries_reported` | int | Số lượng mức lương được báo cáo |
| `location` | object | Địa điểm làm việc |
| `employment_status` | object | Loại hình làm việc (Full-time, Part-time, ...) |
| `job_roles` | object | Nhóm vai trò công việc |

> ⚠️ **Lưu ý:** Dataset này **không có** cột kinh nghiệm (experience/years) và giới tính (gender). Các phân tích được điều chỉnh phù hợp với các cột thực tế trên.

### Quy trình làm sạch dữ liệu

- Chuẩn hóa tên cột (lowercase, loại bỏ khoảng trắng, thay space bằng `_`)
- Xóa dòng trùng lặp (`drop_duplicates`)
- Làm sạch cột `salary`: xử lý định dạng `120k → 120000`, loại ký tự `$`, `,`
- Loại outlier theo phân vị 1%–99%
- Fill missing values: numeric → median, categorical → mode

## 3. 🗺️ HƯỚNG TIẾP CẬN: 4 TẦNG PHÂN TÍCH

Đề tài áp dụng framework **4 tầng Analytics** theo thứ tự từ mô tả → chẩn đoán → dự đoán → đề xuất:

```
🔵 TẦNG 1 – DESCRIPTIVE   "What happened?"
   → Thống kê mô tả (Mean, Median, Std, Skewness, IQR)
   → Phân phối lương: Histogram, KDE, Boxplot
   → Lương theo job_title: Barplot Top 15 + Boxplot nhóm
   → File: 01_descriptive.ipynb

🔵 TẦNG 2 – DIAGNOSTIC    "Why did it happen?"
   → Lương theo employment_status: Bar + Boxplot + ANOVA
   → Lương theo rating: Scatter + Regplot + Pearson correlation
   → Lương theo location: Barplot Top 15 khu vực
   → Lương theo job_roles: Barplot + Boxplot + Heatmap tương quan
   → File: 02_diagnostic.ipynb

🔵 TẦNG 3 – PREDICTIVE    "What will happen?"
   → Huấn luyện & so sánh 7 mô hình Machine Learning
   → Mô hình tốt nhất: XGBoost (R², MAE, MAPE, CV-R²)
   → Phân tích sai số: Actual vs Predicted, Residual plot
   → Feature Importance + SHAP Values (TreeExplainer)
   → File: 03_predictive.ipynb

🔵 TẦNG 4 – PRESCRIPTIVE  "How to make it happen?"
   → Salary Simulator: dự đoán lương 5 mẫu từ test set
   → Salary Roadmap: lộ trình lương theo job_roles
   → Gợi ý hành động: lập trình viên, doanh nghiệp, sinh viên
   → File: 04_prescriptive.ipynb
```

## 7. ⚙️ SETUP & KHÁM PHÁ DỮ LIỆU BAN ĐẦU

````python
# Cài đặt thư viện (chỉ chạy nếu chưa cài)
# !pip install pandas numpy matplotlib seaborn scikit-learn xgboost shap scipy
````

````python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings('ignore')

plt.rcParams.update({
    'figure.figsize'   : (13, 6),
    'font.size'        : 11,
    'axes.titlesize'   : 13,
    'axes.titleweight' : 'bold',
    'axes.spines.top'  : False,
    'axes.spines.right': False,
})
sns.set_style('whitegrid')

print('✅ Import thư viện thành công!')
````

````python
# ── Đọc dữ liệu ─────────────────────────────────────────────
df = pd.read_csv('salary_data.csv')

# ── Chuẩn hóa tên cột ───────────────────────────────────────
df.columns = (
    df.columns
    .str.lower()
    .str.strip()
    .str.replace(' ', '_')
)

# ── Xóa duplicate ───────────────────────────────────────────
df = df.drop_duplicates()

print('✅ Dataset đã sẵn sàng!')
print(f'   Số dòng : {df.shape[0]:,}')
print(f'   Số cột  : {df.shape[1]}')
print(f'\n📌 Danh sách cột:')
for i, col in enumerate(df.columns, 1):
    n_null = df[col].isnull().sum()
    null_str = f'  ⚠️  {n_null} missing' if n_null > 0 else ''
    print(f'   {i:>2}. {col:<25} [{df[col].dtype}]{null_str}')
````

````python
# ── Xem mẫu dữ liệu ─────────────────────────────────────────
print('5 dòng đầu:')
df.head()
````

````python
# ── Thống kê tổng quan ──────────────────────────────────────
print('Thống kê mô tả:')
df.describe(include='all')
````

````python
# ── Kiểm tra missing values ──────────────────────────────────
null_info = df.isnull().sum()
null_info = null_info[null_info > 0]

if len(null_info) == 0:
    print('✅ Không có missing values!')
else:
    print('⚠️  Missing values:')
    for col, n in null_info.items():
        print(f'   {col:<25}: {n:>5} ({n/len(df)*100:.1f}%)')
````

````python
# ── Số lượng giá trị unique mỗi cột ─────────────────────────
print('📌 Số giá trị unique mỗi cột:')
print('=' * 45)

for col in df.columns:
    n_unique = df[col].nunique()
    if df[col].dtype == 'object' and n_unique <= 20:
        vals = ', '.join(str(v) for v in df[col].unique()[:8])
        print(f'   {col:<25}: {n_unique:>4}  → {vals}')
    else:
        print(f'   {col:<25}: {n_unique:>4}')
````
