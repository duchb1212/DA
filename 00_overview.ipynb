## 1. 🎯 VẤN ĐỀ CẦN GIẢI QUYẾT

Lương trong ngành IT luôn là chủ đề được quan tâm nhưng thiếu phân tích dữ liệu cụ thể. Đề tài này nhằm trả lời **4 câu hỏi nghiên cứu** sau:

| # | Câu hỏi | Tầng phân tích |
|---|---------|----------------|
| 1 | Mức lương trung bình của lập trình viên là bao nhiêu? | Descriptive |
| 2 | Các vị trí khác nhau có mức lương khác nhau không? | Descriptive |
| 3 | Có sự chênh lệch lương giữa các nhóm lập trình viên không? | Diagnostic |
| 4 | Yếu tố nào ảnh hưởng mạnh nhất đến lương? | Predictive |

---
## 2. 🗺️ HƯỚNG TIẾP CẬN: 4 TẦNG PHÂN TÍCH

Đề tài áp dụng framework **4 tầng Analytics** theo thứ tự từ mô tả → chẩn đoán → dự đoán → đề xuất:

```
🔵 TẦNG 1 – DESCRIPTIVE   "What happened?"
   → Thống kê mô tả, phân phối lương, so sánh theo vị trí
   → File: 01_descriptive.ipynb

🔵 TẦNG 2 – DIAGNOSTIC    "Why did it happen?"
   → Phân tích nguyên nhân chênh lệch lương giữa các nhóm
   → Kiểm định thống kê (T-test, ANOVA, Pearson)
   → File: 02_diagnostic.ipynb

🔵 TẦNG 3 – PREDICTIVE    "What will happen?"
   → Xây dựng và so sánh 7 mô hình Machine Learning
   → Feature Importance + SHAP Values
   → File: 03_predictive.ipynb

🔵 TẦNG 4 – PRESCRIPTIVE  "How to make it happen?"
   → Đề xuất hành động tối ưu hóa lương
   → File: 04_prescriptive.ipynb
```
